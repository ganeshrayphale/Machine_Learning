```python
import pandas as pd
import numpy as np
```

```python
df= pd.read_csv('covid_toy.csv')
```

```python
df.sample(4)
```

```python
data=df.drop('has_covid',axis=1)
```

```python
data
```

```python
from sklearn.compose import ColumnTransformer
```

```python
from sklearn.preprocessing import OrdinalEncoder
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer
```

```python
#genral syntax is trnasformer=object(trnasfors=[
#tuple('name','transformer','coulmn')]
#remainder=passthough)

transformer=ColumnTransformer(transformers=[
    ('tf1',SimpleImputer(),['fever']),
    ('tf2',OneHotEncoder(sparse_output=False,drop='first'),['gender','city']),
    ('tf3',OrdinalEncoder(categories=[['Mild','Strong']]),['cough'])
],remainder='passthrough')
```

```python
x_train=transformer.fit_transform(data)
```
