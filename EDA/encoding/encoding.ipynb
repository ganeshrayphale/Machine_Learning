one hot encoding

```python
data=pd.read_csv('cars.csv')
```

```python
data.sample(4)
```

```python
data['brand'].value_counts()
```

```python
data['owner'].value_counts()
```

```python
x_train_new=pd.get_dummies(data,columns=['fuel'],drop_first=True,dtype=int)
```

```python
x_train_new
```

```python
counts=data['brand'].value_counts()
```

```python
counts.head(2)
```

```python
threshold=100
repl=counts[counts<=threshold].index
```

```python
x_train_new2=pd.get_dummies(x_train_new['brand'].replace(repl,'others'),drop_first=True,dtype=int)
```

```python
x_train_new2.sample(5)
```

```python
fulltraindata=pd.concat([x_train_new,x_train_new2],axis=1)
```

```python
fulltraindata.sample(5)
```

```python
fulltraindata=fulltraindata.drop('brand',axis=1)
```

```python
fulltraindata.sample(5)
```

```python
fulltraindata.shape
```

```python
fulltraindata['owner'].value_counts()
```

```python
from sklearn.preprocessing import OrdinalEncoder
```

```python
oe=OrdinalEncoder(categories=[['Fourth & Above Owner','Third Owner','Second Owner','First Owner','Test Drive Car']])
```

```python
fulltraindata['owner']=oe.fit_transform(fulltraindata[['owner']])
```

```python
fulltraindata.sample(5)
```
