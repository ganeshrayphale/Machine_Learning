```python
num_cols=df.select_dtypes(include=['int64','float64']).columns
print("Numerical columns:", len(num_cols))
print(num_cols)
```

```python
cat_cols = df.select_dtypes(include=['object', 'category']).columns
print("Categorical columns:", len(cat_cols))
print(cat_cols)
```

```python
threshold = 0.95

dominant_cols = []

for col in cat_cols:
    top_freq = df[col].value_counts(normalize=True, dropna=False).values[0]
    
    if top_freq > threshold:
        dominant_cols.append((col, top_freq))


for col, freq in dominant_cols:
    print(f"{col}: {freq:.2f}")
```

```python
dominant_cols_names = [col for col, freq in dominant_cols]
dominant_cols_names
```

```python
cat_cols = cat_cols.drop(dominant_cols_names)
```
