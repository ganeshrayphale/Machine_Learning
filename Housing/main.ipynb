```python
from sklearn.linear_model import Ridge, Lasso, ElasticNet
from sklearn.model_selection import GridSearchCV

alphas = {'alpha': [0.01, 0.1, 1, 10, 50, 100, 200, 500]}

# Ridge
ridge_grid = GridSearchCV(Ridge(), alphas, cv=5, scoring='r2')
ridge_grid.fit(x_train, y_train)
print("Best Ridge alpha:", ridge_grid.best_params_)
print("Best Ridge CV R2:", round(ridge_grid.best_score_, 4))

# Lasso — add max_iter to fix ConvergenceWarning
lasso_grid = GridSearchCV(Lasso(max_iter=10000), alphas, cv=5, scoring='r2')
lasso_grid.fit(x_train, y_train)
print("\nBest Lasso alpha:", lasso_grid.best_params_)
print("Best Lasso CV R2:", round(lasso_grid.best_score_, 4))

# ElasticNet — also add max_iter
en_params = {
    'alpha':    [0.01, 0.1, 1, 10, 50],
    'l1_ratio': [0.2, 0.5, 0.7, 0.9]
}
en_grid = GridSearchCV(ElasticNet(max_iter=10000), en_params, cv=5, scoring='r2')
en_grid.fit(x_train, y_train)
print("\nBest ElasticNet params:", en_grid.best_params_)
print("Best ElasticNet CV R2:", round(en_grid.best_score_, 4))

# Final evaluation on test set with best models
from sklearn.metrics import r2_score, mean_absolute_error

for name, grid in [('Ridge', ridge_grid), ('Lasso', lasso_grid), ('ElasticNet', en_grid)]:
    y_pred = grid.best_estimator_.predict(x_test)
    print(f"\n{name} Test | MAE: ${mean_absolute_error(y_test, y_pred):,.0f} | R2: {r2_score(y_test, y_pred):.4f}")
```
