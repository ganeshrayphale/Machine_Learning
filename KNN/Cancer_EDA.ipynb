```python
X = df.drop("diagnosis_M", axis=1)
y = df["diagnosis_M"]
```

```python
from sklearn.model_selection import train_test_split

X_train,X_test,y_train,y_test=train_test_split(
    X,y,test_size=0.2,random_state=2,stratify=y
)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
print(X_train_scaled.shape)
print(X_test_scaled.shape)
print(y_train.shape)
print(y_test.shape)
```

```python
import joblib

joblib.dump(scaler, "scaler.pkl")
```

```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
knn=KNeighborsClassifier(n_neighbors=5)
```

```python
X_train_scaled.shape
```

```python
knn.fit(X_train_scaled,y_train)
```

```python
y_pred=knn.predict(X_test_scaled)
```

```python
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(y_test, y_pred)
print("Accuracy:", accuracy)
```

```python
from sklearn.metrics import classification_report

print(classification_report(y_test, y_pred))
```

```python
scores=[]
for i in range(1,20):
    knn=KNeighborsClassifier(n_neighbors=i)
    knn.fit(X_train_scaled,y_train)
    y_pred=knn.predict(X_test_scaled)
    scores.append(accuracy_score(y_test,y_pred))
```

```python
scores
```

```python
import matplotlib.pyplot as plt
plt.plot(range(1,20),scores)
```

```python
from sklearn.model_selection import GridSearchCV

param_grid = {"n_neighbors": range(1, 21)}

grid = GridSearchCV(
    KNeighborsClassifier(),
    param_grid,
    cv=10,
    scoring="accuracy"
)

grid.fit(X_train_scaled, y_train)

print("Best K:", grid.best_params_)
print("Best CV Accuracy:", grid.best_score_)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import ListedColormap
from sklearn.neighbors import KNeighborsClassifier
from sklearn.preprocessing import StandardScaler

# 1. Select two features to visualize the 2D decision boundary
# (Using 'radius_mean' and 'texture_mean' from your dataset)
features = ['radius_mean', 'texture_mean']
X_2d = df[features].values
y = df['diagnosis_M'].values

# 2. Scale the features (since KNN is highly sensitive to feature scales)
scaler_2d = StandardScaler()
X_2d_scaled = scaler_2d.fit_transform(X_2d)

# 3. Fit a KNN model on these two features 
# (Using K=9 which was found to be the best K via your GridSearchCV)
knn_2d = KNeighborsClassifier(n_neighbors=9)
knn_2d.fit(X_2d_scaled, y)

# 4. Define boundary limits and create a dense meshgrid
x_min, x_max = X_2d_scaled[:, 0].min() - 0.5, X_2d_scaled[:, 0].max() + 0.5
y_min, y_max = X_2d_scaled[:, 1].min() - 0.5, X_2d_scaled[:, 1].max() + 0.5
xx, yy = np.meshgrid(np.arange(x_min, x_max, 0.02),
                     np.arange(y_min, y_max, 0.02))

# 5. Predict the class for each point on the grid to map out the boundary
Z = knn_2d.predict(np.c_[xx.ravel(), yy.ravel()])
Z = Z.reshape(xx.shape)

# 6. Setup premium-looking color maps (soft background, bold points)
cmap_light = ListedColormap(['#FFCCCC', '#CCE5FF'])  # Light Red (Benign), Light Blue (Malignant)
cmap_bold = ['#FF3333', '#0066CC']                  # Bold Red, Bold Blue

# 7. Plot the decision boundary surface
plt.figure(figsize=(10, 8), dpi=100)
plt.contourf(xx, yy, Z, alpha=0.6, cmap=cmap_light)

# 8. Overlay the scatter plot of actual data points
for class_value, color in zip(np.unique(y), cmap_bold):
    label_name = 'Malignant' if class_value == 1 else 'Benign'
    plt.scatter(
        X_2d_scaled[y == class_value, 0],
        X_2d_scaled[y == class_value, 1],
        c=color,
        label=label_name,
        edgecolors='k',
        alpha=0.8,
        s=40
    )

# 9. Format and style the plot
plt.title('KNN Decision Boundary (K=9) on Scaled Features', fontsize=14, fontweight='bold', pad=15)
plt.xlabel(f'Scaled {features[0]}', fontsize=12)
plt.ylabel(f'Scaled {features[1]}', fontsize=12)
plt.legend(loc='best', frameon=True, facecolor='white', edgecolor='none')
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```
